# Notebook n°4 : Machine Learning: Prédiction des retards sans graphe

### Problématique

> **Dans quelle mesure la structure du réseau TGV est-elle associée à la régularité des liaisons, et peut-on exploiter cette structure pour améliorer la prédiction des retards ?**

## 1. Objectifs de ce notebook

Après avoir étudié la qualité des données dans le notebook 01 et la régularité des liaisons dans le notebook 02, nous allons maintenant chercher à **prédire les retards des TGV à l'aide de modèles de Machine Learning**.

Dans ce premier notebook de modélisation, nous utiliserons uniquement les caractéristiques classiques des liaisons, sans prendre en compte les indicateurs issus de l'analyse du graphe.

L'objectif est de construire un **modèle de référence (*baseline*)**, qui nous permettra ensuite de mesurer si l'ajout des caractéristiques du réseau TGV améliore les prédictions.

Nous allons notamment :

1. Définir la variable à prédire et les variables explicatives.
2. Préparer les données pour la modélisation.
3. Séparer chronologiquement les données d'entraînement, de validation et de test.
4. Construire un modèle naïf de référence.
5. Entraîner et comparer plusieurs modèles de Machine Learning.
6. Évaluer leurs performances et analyser leurs erreurs.

Les résultats obtenus serviront de référence pour le notebook 05, dans lequel nous ajouterons les indicateurs du graphe afin d'évaluer leur apport à la prédiction des retards.

## 2. Méthodologie

Nous chercherons à prédire le taux de retard d'une liaison TGV pour un mois donné, à partir des informations disponibles avant ce mois.

Nous privilégierons une séparation chronologique des données afin de reproduire une situation de prédiction réelle et d'éviter d'utiliser des informations futures lors de l'entraînement.

Le choix définitif de la variable cible et des variables explicatives sera effectué après validation des données préparées par l'ETL.

Enfin, nous veillerons à utiliser le même protocole d'évaluation dans les notebooks 04 et 05, pour que la comparaison entre les modèles avec et sans graphe soit fiable.
